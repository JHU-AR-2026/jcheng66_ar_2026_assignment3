```python
import numpy as np

def camera_to_pixel(Pc, W, H, n, f, fx, fy, ox, oy):
    # OpenGL projection matrix
    M_proj = np.array([
        [2*fx/W, 0, (W - 2*ox)/W, 0],
        [0, 2*fy/H, (2*oy - H)/H, 0],
        [0, 0, -(f+n)/(f-n), -2*f*n/(f-n)],
        [0, 0, -1, 0]
    ], dtype=float)

    P_clip = M_proj @ Pc

    P_ndc = P_clip[:3] / P_clip[3]

    u = 0.5 * (P_ndc[0] + 1) * W
    v = 0.5 * (P_ndc[1] + 1) * H

    return M_proj, P_clip, P_ndc, (u, v)
```

```python
# Testing
W, H = 800, 600
n, f = 0.2, 20
fx, fy = 800, 800
ox, oy = 400, 400

Pc = np.array([0.3, -0.15, -6.0, 1.0])

M, P_clip, P_ndc, pixel = camera_to_pixel(
    Pc, W, H, n, f, fx, fy, ox, oy
)

np.set_printoptions(precision=4, suppress=True)

print("Projection matrix:\n", M)
print("Clip coordinates:", P_clip)
print("NDC coordinates:", P_ndc)
print("Pixel coordinates:", pixel)
```
